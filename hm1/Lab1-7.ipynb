8. Выведите минимальное и максимальное представимое значение для каждого скалярного типа numpy

````python
all_types = [
    np.int8, np.int16, np.int32, np.int64, np.longlong,
    np.uint8, np.uint16, np.uint32, np.uint64, np.ulonglong,
    np.float16, np.float32, np.float64, np.longdouble,
    np.bool_
]


for i in all_types:
    print(f"class type: {i}")
    if (i == np.bool_):
        print("min = False, max = True")
    elif issubclass(i, (np.unsignedinteger, np.signedinteger)):
        info = np.iinfo(i)
        print(f"min = {info.min}, max = {info.max}")
    else:
        info = np.finfo(i)
        print(f"min = {info.min}, max = {info.max}")
````

3. Правда ли, что люди с высоким уровнем стресса имеют более высокое кровяное давление, чем те, кто не подвержен стрессу

````python
stress_median = df['Stress Levels'].median()

df["High Blood Pressure"] = df["Blood Pressure (s/d)"].str.split("/").str[0].astype(float)

high = df[df['Stress Levels'] > stress_median]
low = df[df['Stress Levels'] < stress_median]

mean_high = high["High Blood Pressure"].mean()
mean_low = low["High Blood Pressure"].mean()

print(f"Medium stress level: {stress_median:.1f}")

print(f"people with higher stress level have higher blood pressure: {mean_high > mean_low}")
````

8. Правда ли, что у людей с индексом массы тела больше 75го перцентиля встречаются хронические заболевания чаще, чем у людей с индесом массы тела в переделах от 30 до 60 перцентиля?

````python
bmi30 = df["BMI"].quantile(0.30)
bmi60 = df["BMI"].quantile(0.60)
bmi75 = df["BMI"].quantile(0.75)

high = df[df["BMI"] > bmi75]
low = df[df["BMI"] > bmi30]
low = low[low["BMI"] < bmi60]


cd_h = high[high["Chronic Diseases"] != "Nan"]
cd_l = low[low["Chronic Diseases"] != "Nan"]

res_h = cd_h["Chronic Diseases"].value_counts()
res_l = cd_l["Chronic Diseases"].value_counts()


print(f"chronic diseases of people with high BMI: {sum(res_h)}")
print(f"chronic diseases of people with low BMI: {sum(res_l)}")

print(f"Is it true that people with a body mass index above the 75th percentile are more likely to have chronic diseases than people with a body mass index between the 30th and 60th percentile: {sum(res_h) > sum(res_l)}")
````
